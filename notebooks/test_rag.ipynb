```python
# loading pdf files 
# Cell 4: Read PDF files (PDF LOADING PART)
def read_pdf_file(pdf_path):
    """Simple function to read text from PDF"""
    text = ""
    with open(pdf_path, 'rb') as file:
        pdf_reader = PyPDF2.PdfReader(file)
        for page in pdf_reader.pages:
            text += page.extract_text() + "\n"
    return text

# Read all PDFs from the correct path
pdf_folder = "../datas"  # Use the correct path that worked
pdf_files = [f for f in os.listdir(pdf_folder) if f.endswith('.pdf')]

all_texts = []
for pdf_file in pdf_files:
    pdf_path = os.path.join(pdf_folder, pdf_file)
    print(f"Reading: {pdf_file}")
    text = read_pdf_file(pdf_path)
    all_texts.append({"name": pdf_file, "text": text})
    print(f"   - Read {len(text)} characters")

print(f"\n✅ Successfully loaded {len(all_texts)} PDF files")
```

```python
# Cell 5: Split text into chunks
print("Splitting text into chunks...")

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)

all_chunks = []
for doc in all_texts:
    chunks = splitter.split_text(doc["text"])
    for i, chunk in enumerate(chunks):
        all_chunks.append({
            "text": chunk,
            "source": doc["name"],
            "chunk_id": i
        })
    print(f"📑 {doc['name']}: {len(chunks)} chunks")

print(f"\n📊 Total chunks created: {len(all_chunks)}")
```
